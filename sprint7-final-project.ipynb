```python
# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int)
usage["is_call"] = (usage["type"] == "call").astype(int)

# Agrupar información por usuario
usage_agg = usage.groupby("user_id").agg(
    cant_mensajes=("is_text", "sum"),
    cant_llamadas=("is_call", "sum"),
    cant_minutos_llamada=("duration", "sum")
).reset_index()

# Observar resultado
usage_agg.head(3)
```

```python
# Renombrar columnas
usage_agg = usage_agg.rename(columns={
    "cant_mensajes": "cant_mensajes",
    "cant_llamadas": "cant_llamadas",
    "cant_minutos_llamada": "cant_minutos_llamada"
})

# Observar resultado
usage_agg.head(3)
```

```python
# Unir usuarios con sus métricas de uso
user_profile = users.merge(usage_agg, on="user_id", how="left")

# Completar con 0 las métricas de usuarios sin registros de actividad
columnas_uso = [
    "cant_mensajes",
    "cant_llamadas",
    "cant_minutos_llamada"
]

user_profile[columnas_uso] = user_profile[columnas_uso].fillna(0)

# Verificar resultados
print("Dimensiones de user_profile:", user_profile.shape)

print("\nValores faltantes:")
print(user_profile[columnas_uso].isna().sum())

display(user_profile.head())
```

```python
# Calcular cantidad y porcentaje de valores atípicos

columnas = [
    'cant_mensajes',
    'cant_llamadas',
    'cant_minutos_llamada'
]

for col in columnas:
    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    atipicos = (
        (user_profile[col] < limite_inferior) |
        (user_profile[col] > limite_superior)
    )

    cantidad = atipicos.sum()
    porcentaje = (cantidad / len(user_profile)) * 100

    print(f'\n{col}:')
    print(f'Cantidad de valores atípicos: {cantidad}')
    print(f'Porcentaje de usuarios: {porcentaje:.2f}%')
```

Análisis de valores atípicos

Mediante el método del rango intercuartílico (IQR) se identificaron valores atípicos en las variables de actividad de los usuarios. En la cantidad de mensajes se encontraron 46 registros (1,15 %), en la cantidad de llamadas 30 registros (0,75 %) y en los minutos de llamadas 109 registros (2,73 %).

Por esta razón, se conservaron para el análisis, ya que no existe evidencia suficiente para considerarlos errores en los datos.

```python

# Calcular límites con el método IQR
columnas_limites = [
    'cant_mensajes',
    'cant_llamadas',
    'cant_minutos_llamada'
]

for col in columnas_limites:
    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    print(f'\nVariable: {col}')
    print(f'Límite inferior: {limite_inferior:.2f}')
    print(f'Límite superior: {limite_superior:.2f}')
```
